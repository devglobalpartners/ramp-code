```python
import pandas as pd
import os
```

```python
# Load your size CSV
df = pd.read_csv("/home/jovyan/work/my_scripts/Ana_missing_files.csv")
```

```python
df.columns=["filename","sizeGB"]
```

```python
# Sort to get a predictable order
df = df.sort_values("sizeGB", ascending=False)
df["filename"]=df["filename"].astype(str)
```

```python
df
```

```python

batches = []
current_batch = []
current_size = 0
batch_id = 1

for _, row in df.iterrows():
    if current_size + row["sizeGB"] <= 120:
        current_batch.append(row)
        current_size += row["sizeGB"]
    else:
        size_total=pd.DataFrame(current_batch)["sizeGB"].sum()
        print(f"size of batch:{size_total}")
        
        batches.append(pd.DataFrame(current_batch))
        current_batch = [row]
        current_size = row["sizeGB"]
        batch_id += 1

        

# Add the last batch
if current_batch:
    batches.append(pd.DataFrame(current_batch))
    
print(f"Created {len(batches)} batches")

```

```python
# Save each batch to separate CSVs
for i, batch in enumerate(batches, 1):
    batch.to_csv(f"/home/jovyan/work/batches_tmp/batch_{i:02d}.csv", index=False)

print(f"Created {len(batches)} batches.")
```
