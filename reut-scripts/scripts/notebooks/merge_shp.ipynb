```python
import geopandas as gpd
import pandas as pd
import os
```

```python
shapefile_dir = ("/home/jovyan/work/geometries/downloaded_tiles/shapefiles")
# List all shapefiles in the directory
shapefiles = [os.path.join(shapefile_dir, f) 
              for f in os.listdir(shapefile_dir) 
              if f.endswith('.shp')]

# Read and combine all shapefiles into one GeoDataFrame
gdfs = [gpd.read_file(shp) for shp in shapefiles]

# Concatenate them into a single GeoDataFrame
merged_gdf = gpd.GeoDataFrame(pd.concat(gdfs, ignore_index=True), crs=gdfs[0].crs)
merged_gdf.plot()
# (Optional) Save merged file
merged_gdf.to_file("/home/jovyan/work/geometries/downloaded_tiles/merged_output2.gpkg")
```
