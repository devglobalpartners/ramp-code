```python
import os
from concurrent.futures import ThreadPoolExecutor
import rasterio

input_folder = "/home/jovyan/work/imagery/Nepal/08122025/4326_remove_nir"
output_folder = "/home/jovyan/work/imagery/Nepal/08122025/4326_remove_nir_rgb"
os.makedirs(output_folder, exist_ok=True)

# Gather lists of source and destination TIFF files
source_files = [f for f in os.listdir(input_folder) if f.lower().endswith(".tif")]
dest_files = set(f for f in os.listdir(output_folder) if f.lower().endswith(".tif"))

print(f"Source files: {len(source_files)}")
print(f"Destination files: {len(dest_files)}")

# Only process files that are not yet in the destination
files_to_process = [f for f in source_files if f not in dest_files]
print(f"Files to process: {len(files_to_process)}")

def convert_bgr_to_rgb(filename):
    in_path = os.path.join(input_folder, filename)
    out_path = os.path.join(output_folder, filename)

    with rasterio.open(in_path) as src:
        img = src.read()  # (C, H, W)
        profile = src.profile

    rgb = img[[2, 1, 0], :, :]  # Swap BGR → RGB
    profile.update(count=3)

    with rasterio.open(out_path, "w", **profile) as dst:
        dst.write(rgb)

    print(f"Saved RGB file: {out_path}")

# Increase the number of threads for faster processing
num_workers = 8
with ThreadPoolExecutor(max_workers=num_workers) as executor:
    executor.map(convert_bgr_to_rgb, files_to_process)
```

```python
import os
import rasterio

input_folder = "/home/jovyan/work/imagery/Nepal/merged_04122025/imgs_copied"
output_folder = "/home/jovyan/work/imagery/Nepal/merged_04122025/imgs_bgr_to_rgb"

os.makedirs(output_folder, exist_ok=True)

for filename in os.listdir(input_folder):
    if filename.lower().endswith(".tif"):
        in_path = os.path.join(input_folder, filename)
        out_path = os.path.join(output_folder, filename)

        with rasterio.open(in_path) as src:
            img = src.read()  # (C, H, W)
            profile = src.profile

        # Swap BGR → RGB
        rgb = img[[2, 1, 0], :, :]

        # Update profile for 3-band output
        profile.update(count=3)

        with rasterio.open(out_path, "w", **profile) as dst:
            dst.write(rgb)

        print(f"Saved RGB file: {out_path}")
```
