```python
import pandas as pd
import matplotlib.pyplot as plt

# Load your CSV file
history = pd.read_csv("/home/jovyan/work/training-models/south_afghanistan/v3/results/history_live.csv")

# Create figure with 2 subplots
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

# ----- Accuracy -----
ax1.plot(history["sparse_categorical_accuracy"], label="Train Accuracy")
ax1.plot(history["val_sparse_categorical_accuracy"], label="Val Accuracy")

ax1.set_title("Accuracy")
ax1.set_xlabel("Epoch")
ax1.set_ylabel("Accuracy")
ax1.legend()
ax1.grid(True)

# ----- Loss -----
ax2.plot(history["loss"], label="Train Loss")
ax2.plot(history["val_loss"], label="Val Loss")

ax2.set_title("Loss")
ax2.set_xlabel("Epoch")
ax2.set_ylabel("Loss")
ax2.legend()
ax2.grid(True)

plt.tight_layout()
plt.show()
```
